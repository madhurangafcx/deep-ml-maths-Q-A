# Reshape Matrix

Write a Python function that reshapes a given matrix into a specified shape. if it cant be reshaped return back an empty list []

Example:
Input:
a = [[1,2,3,4],[5,6,7,8]], new_shape = (4, 2)

Output:
[[1, 2], [3, 4], [5, 6], [7, 8]]

Reasoning:
The given matrix is reshaped from 2x4 to 4x2.

```python
import numpy as np

def reshape_matrix(a: list[list[int|float]], new_shape: tuple[int, int]) -> list[list[int|float]]:
	# Convert the standerd Python list to NumPy array
	np_matrix = np.array(a)

	try:
		# Attemp to reshape the array
		# If the diamentions are incompatible, this will trigger Value error
		reshaped_array = np_matrix.reshape(new_shape)

		# Convet back to standerd Python list as requested
		reshaped_matrix = reshaped_array.tolist()
		return reshaped_matrix

	except ValueError:
		# Catch the error if dimentions don't match and return an empty list
		return []
```

```python
print(reshape_matrix([[1, 2, 3], [4, 5, 6]], (2, 3)))
```
