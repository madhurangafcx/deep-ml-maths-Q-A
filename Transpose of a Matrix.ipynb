# Transpose of a Matrix

Write a Python function that computes the transpose of a given 2D matrix. The transpose of a matrix is formed by turning its rows into columns and columns into rows. For an m×n matrix, the transpose will be an n×m matrix.

Example:
Input:
a = [[1, 2, 3], [4, 5, 6]]

Output:
[[1, 4], [2, 5], [3, 6]]

Reasoning:
The input is a 2×3 matrix. The transpose swaps rows and columns: the first row [1, 2, 3] becomes the first column, and the second row [4, 5, 6] becomes the second column, resulting in a 3×2 matrix.

```python
def transpose_matrix(a: list[list[int|float]]) -> list[list[int|float]]:
    #handle empty matix test edge cases
    if not a:
        return []

    rows = len(a)
    cols = len(a[0])

    result = []

    # outer loop interate through colums of the origial matrix
    for i in range(cols):
        new_row = []
        # inner loop iterates through rows of the original matrix
        for j in range(rows):
            new_row.append(a[j][i])


        # add the completed new row to the transposed  matrix
        result.append(new_row)

    return result
```

```python
print(transpose_matrix([[1,2],[3,4],[5,6]]))  # -> [[1,3,5],[2,4,6]]
```
